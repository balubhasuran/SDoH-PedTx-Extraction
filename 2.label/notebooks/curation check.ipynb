```python
import glob
import pandas as pd
import json
```

```python
info_dict = {}
for i in glob.glob('../data/processed_data/*.json'):
    file = json.load(open(i,'r'))
    name = i.split("/")[-1].split(".json")[0]
    info_dict[name] = []
    for idx, info in file.items():
        if "SDoH" in info.keys():
            sdohs = info['SDoH']
            for j in sdohs:
                if "Mental Health" in j.keys():
                    info_dict[name].append(idx)
                    break
```

```python
sorted_dict_by_key = dict(sorted(info_dict.items()))
```
