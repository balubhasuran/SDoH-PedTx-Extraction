```python
import pandas as pd
import json
import glob
```

```python
df = pd.read_csv('../data/SDoH_data_columns.csv',header=None)
```

```python
final_output = {}
for i in df.iterrows():
    info = i[1]
    key = info[0]
    value = []
    for j in info[1:]:
        if not pd.isna(j):
            value.append(j)
    final_output[key]=value
    
```

```python
files = glob.glob('../data/processed_data/*.json')
```

```python
all_possibles = {}
for i in files:
    file = json.load(open(i,'r'))
    for k,v in file.items():
        if 'SDoH' in v.keys():
            sdoh = v['SDoH']
            for s in sdoh:
                for labels in s.values():
                    for k2,v2 in labels.items():
                        if k2 not in all_possibles.keys():
                            all_possibles[k2] = set([])
                        all_possibles[k2].add(v2)
        else:
            continue
```
