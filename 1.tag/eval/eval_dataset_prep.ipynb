```python
import pandas as pd
import json
from glob import glob
```

```python
true_data_path = glob("../data/splitted_data/fold_*/test.json")
pred_data_path = glob("../output/roberta_fold_*_lr_5e-5/predictions/*.txt")
```

```python
output_list = []
for i, j in zip(true_data_path, pred_data_path):
    true = json.load(open(i,"r"))
    pred = open(j,'r').readlines()
    for trues,preds in zip(true, pred):
        sentence = " ".join(trues['tokens'])
        true_tags = set([x.split("-")[1] for x in trues['ner_tags'] if x != "O"])
        pred_tags = set([x.split("-")[1] for x in preds.rstrip().split() if x != "O"])
        if len(true_tags - pred_tags) == 0:
            correct = True
        else:
            correct = False
        output_list.append([sentence, ",".join(list(true_tags)), ",".join(list(pred_tags)), correct])
```

```python
eval_data = pd.DataFrame(output_list, columns = ["sentence", "true", "pred", "match"])
```

```python
eval_data.to_csv("./eval_data_all.csv")
```

```python
wrong_data = eval_data[eval_data.match==False]
```

```python
wrong_data.to_csv("./eval_data_wrong.csv")
```
