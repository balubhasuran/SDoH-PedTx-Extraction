```python
import json
import numpy as np
from glob import glob
from pathlib import Path

# If you haven't run your previous cell, re-initialize your paths here
true_paths = sorted([Path(p).as_posix() for p in glob("../../2. label/data/processed_data/*.json")])

document_word_counts = []

for filepath in true_paths:
    with open(filepath, 'r', encoding='utf-8') as f:
        data = json.load(f)
        
    doc_word_count = 0
    
    # Based on your previous JSON structure where data is a dictionary of items
    for idx, item in data.items():
        sentence = item.get("sentence", "")
        # Splitting by whitespace is the standard way to count words for clinical text
        words = sentence.split()
        doc_word_count += len(words)
        
    document_word_counts.append(doc_word_count)

# Calculate and print the statistics
if document_word_counts:
    avg_words = np.mean(document_word_counts)
    median_words = np.median(document_word_counts)
    min_words = np.min(document_word_counts)
    max_words = np.max(document_word_counts)
    
    print("=== Document Length Statistics ===")
    print(f"Total documents processed: {len(document_word_counts)}")
    print(f"Average word count:        {avg_words:.1f} words")
    print(f"Median word count:         {median_words:.1f} words")
    print(f"Length range:              {min_words} to {max_words} words")
else:
    print("No files were found. Please check your glob path.")
```
