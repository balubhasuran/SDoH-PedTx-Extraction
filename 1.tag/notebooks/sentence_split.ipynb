```python
cnt_dict = {}
train_file = "../data/splitted_data/sentence_split/train.json"
doc_id = train_file.split("_")[0]
file = json.load(open(train_file, 'r'))
for i in file:
    ner_tags = i['ner_tags']
    for ner_tag in ner_tags:
        if ner_tag.startswith("B"):
            tag = ner_tag.split('-')[1]
            if tag not in cnt_dict.keys():
                cnt_dict[tag] = 1
            else:
                cnt_dict[tag] += 1
```
