```python
import pandas as pd
import json
from glob import glob
```

```python
data_path = glob("../data/splitted_data/fold_*/*.json")
```

```python
def generate_ctran_dataset(input_path):
    data = json.load(open(input_path,"r",encoding="utf-8"))
    output_path = input_path.replace("json","iob")
    data_list = []
    for i in data:
        tokens = i['tokens']
        tags = i['ner_tags']
        tokens_str = " ".join(tokens)
        tokens_str = "BOS " + tokens_str + " EOS"
        tag_str = " ".join(tags)
        tag_str = "O " + tag_str + " O"
        final_str = tokens_str + "\t" + tag_str
        data_list.append(final_str)
    with open(output_path,'w',encoding="utf-8") as f:
        for i in data_list:
            f.write(i)
            f.write("\n")
    print(f"{output_path} generated")
```

```python
for i in data_path:
    generate_ctran_dataset(i)
```

```python
train_data = json.load(open("../data/splitted_data/sentence_split/train.json",'r'))
test_data = json.load(open("../data/splitted_data/sentence_split/test.json",'r'))
```

```python
train_list = []
for i in train_data:
    tokens = i['tokens']
    tags = i['ner_tags']
    tokens_str = " ".join(tokens)
    tokens_str = "BOS " + tokens_str + " EOS"
    tag_str = " ".join(tags)
    tag_str = "O " + tag_str + " O"
    final_str = tokens_str + "\t" + tag_str
    train_list.append(final_str)
```

```python
test_list = []
for i in test_data:
    tokens = i['tokens']
    tags = i['ner_tags']
    tokens_str = " ".join(tokens)
    tokens_str = "BOS " + tokens_str + " EOS"
    tag_str = " ".join(tags)
    tag_str = "O " + tag_str + " O"
    final_str = tokens_str + "\t" + tag_str
    test_list.append(final_str)
```

```python
with open("../data/splitted_data/sentence_split/ctran_train.iob",'w') as f:
    for i in train_list:
        f.write(i)
        f.write("\n")
```

```python
with open("../data/splitted_data/sentence_split/ctran_test.iob",'w') as f:
    for i in test_list:
        f.write(i)
        f.write("\n")
```
