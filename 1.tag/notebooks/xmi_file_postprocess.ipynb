```python
import glob
from tqdm import tqdm
```

```python
file_list = glob.glob("../data/xmi_data/*")
```

```python
for i in tqdm(file_list):
    file_name = i.split("\\")[1].split('.')[0]
    with open(f"../data/xmi_data_new/{file_name}_new.xmi", 'w') as f:
        file = open(i,'r').readlines()
        for line in file:
            if "custom:SDoHlevel2" in line:
                continue
            else:
                f.writelines(line)
    f.close()

                
```
