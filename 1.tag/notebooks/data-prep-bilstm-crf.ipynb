```python
import pandas as pd
import json
from glob import glob
```

```python
data_path = glob("../data/splitted_data/fold_*/*.json")
```

```python
def generate_bilstm_dataset(input_path):
    
    data = json.load(open(input_path,"r",encoding="utf-8"))
    output_path = input_path.replace("json","txt")
    if "train" in input_path:
        data_list = []
        for i in data:
            tokens = i['tokens']
            tags = i['ner_tags']
            txt = ""
            for v1,v2 in zip(tokens,tags):
                txt += f"{v1}/{v2} "
            txt = txt.rstrip()
            data_list.append(txt)
    elif "test" in input_path:
        data_list = []
        for i in data:
            tokens = i['tokens']
            txt = ""
            for token in tokens:
                txt += f"{token} "
            txt = txt.rstrip()
            data_list.append(txt)
        
    with open(output_path,'w',encoding="utf-8") as f:
        for i in data_list:
            f.write(i)
            f.write("\n")
    print(f"{output_path} generated")
```

```python
for i in data_path:
    generate_bilstm_dataset(i)
```
